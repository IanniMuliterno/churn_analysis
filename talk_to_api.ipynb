```python
import io
import pandas as pd
import requests
host = 'http://127.0.0.1:8000/'
```

```python
df = pd.read_csv('ecom-user-churn-data.csv')
test_df = df.sample(10)
test_output = test_df['target_class']
test_df.drop('target_class',axis=1,inplace=True)
```

```python
requests.get(host + '/').text
```

```python
csv_buffer = io.StringIO()
test_df.to_csv(csv_buffer, index=False)
csv_buffer.seek(0)

files = {"file": ("test.csv", csv_buffer, "text/csv")}
%time response = requests.post(host + "predict", files=files)
```
